```python
price = float(input("Enter item price: "))
quantity = int(input("Enter quantity: "))

subtotal = price * quantity
gst = subtotal * 18 / 100

if subtotal < 500:
    delivery = 40
else:
    delivery = 0

total = subtotal + gst + delivery

print("Subtotal: ₹", round(subtotal, 2))
print("GST (18%): ₹", round(gst, 2))
print("Delivery: ₹", delivery)
print("Total: ₹", round(total, 2))
```

```python
temp = float(input("Enter temperature: "))
unit = input("Enter unit (C/F): ").upper()

if unit == "C":
    result = (temp * 9/5) + 32
    print("Temperature in Fahrenheit:", round(result, 2))

elif unit == "F":
    result = (temp - 32) * 5/9
    print("Temperature in Celsius:", round(result, 2))

else:
    print("Invalid unit! Please enter C or F.")
```

```python
x = input("Enter a value: ")
digits = 0
dots = 0

for ch in x:
    if ch >= '0' and ch <= '9':
        digits += 1
    elif ch == '.':
        dots += 1

if digits == len(x):
    print("Type: int")
elif digits + dots == len(x) and dots == 1:
    print("Type: float")
else:
    print("Type: str")
```

```python
income = int(input("Enter annual income: "))

if income <= 300000:
    tax = 0

elif income <= 700000:
    tax = (income - 300000) * 5 / 100

elif income <= 1000000:
    tax = 20000 + (income - 700000) * 10 / 100

elif income <= 1200000:
    tax = 50000 + (income - 1000000) * 15 / 100
else:
    tax = 140000 + (income - 1500000) * 30 / 100

print("Annual Income: ₹", income)
print("Total Income Tax: ₹", tax)
```

```python
runs = int(input("Enter runs: "))
balls = int(input("Enter balls: "))

if balls == 0:
    print("Cannot calculate strike rate")
else:
    st_rt= runs / balls * 100
    if st_rt> 150:
        print("Explosive")
    elif st_rt > 100:
        print("Aggressive")
    elif st_rt >= 75:
        print("Steady")
    else:
        print("Defensive")
```

```python
w = float(input("Weight: "))
h = float(input("Height in cm: "))

h = h / 100
bmi = w / (h * h)

print("BMI:", bmi)

if bmi < 18.5:
    print("Underweight")
elif bmi <= 22.9:
    print("Normal")
elif bmi <= 24.9:
    print("Overweight")
elif bmi <= 29.9:
    print("Obese Class I")
else:
    print("Obese Class II")
```

```python
num = int(input("Enter a number: "))
org = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reverse:", reverse)

if org == reverse and org >= 0:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
s = input("Enter a sentence: ")

v = 0
c = 0
d = 0
sp = 0
special = 0

for ch in s:
    if ch in "aeiouAEIOU":
        v += 1
    elif ch >= "a" and ch <= "z":
        c += 1
    elif ch >= "0" and ch <= "9":
        d += 1
    elif ch == " ":
        sp += 1
    else:
        special += 1

print("Characters:", len(s))
print("Words:", len(s.split()))
print("Vowels:", v)
print("Consonants:", c)
print("Digits:", d)
print("Spaces:", sp)
print("Special characters:", special)
```

```python
states = {
    "Tamil Nadu": "Chennai",
    "Kerala": "Kochi",
    "Karnataka": "Bengaluru",
    "Maharashtra": "Mumbai",
    "Telangana": "Hyderabad"
}

while True:
    print("1.Add  2.Update  3.Delete")
    print("4.Search  5.Display  6.Exit")

    ch = int(input("Enter choice: "))

    if ch == 1:
        s = input("State: ")
        c = input("Capital: ")
        states[s] = c

    elif ch == 2:
        s = input("State: ")
        states[s] = input("New capital: ")

    elif ch == 3:
        s = input("State: ")
        if s in states:
            del states[s]

    elif ch == 4:
        s = input("State: ")
        print(states.get(s, "Not found"))

    elif ch == 5:
        print(states)

    elif ch == 6:
        break
```
